# Colab runtime for the ensemble experiment

Open this file, click **Select Kernel** (top right) and choose **Colab**.
Once it says connected, tell Claude and it will drive the rest from here.

Why: XGBoost at `max_depth=12` needs roughly 3-4 GB and this machine has 1.5 GB
free. Colab gives ~13 GB plus an optional GPU, which also turns a 24-minute
training run into a couple of minutes.

```python
# Sanity check - run this after connecting to confirm which machine you are on.
import os, platform, subprocess

print("platform :", platform.platform())
print("cpus     :", os.cpu_count())
print("colab    :", os.path.exists("/content"))

try:
    import psutil
    print("ram      : %.1f GB" % (psutil.virtual_memory().total / 1e9))
except ImportError:
    pass

gpu = subprocess.run(
    ["nvidia-smi", "--query-gpu=name,memory.total", "--format=csv,noheader"],
    capture_output=True, text=True,
)
print("gpu      :", gpu.stdout.strip() or "none")
```
