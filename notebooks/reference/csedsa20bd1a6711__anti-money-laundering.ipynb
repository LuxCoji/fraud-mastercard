# Dealing with big-data (simple demo)

This small project aims to apply an approach for importing and processing a large volume of data. The demonstrative goal is to calculate the total received and paid for all currencies in the database and display a table with the total value received and paid for each currency in the entire data.

To view the database, the "nrows=5_000" parameter was used, which displays the first 5,000 rows of the database.

And to perform processing with the entire database, the database'll be read in chunks:
> for i, chunk in enumerate(pd.read_csv(source, chunksize=10_000))

Database: HI-Large_Trans.csv (17.05 GB)<br />
Runtime: ~10 min

##### Additional information:

But it’s not just the type or amount of data that’s important, it’s what organizations do with the data that matters. Big data can be analyzed for insights that improve decisions and give confidence for making strategic business moves.](https://www.sas.com/en_us/insights/big-data/what-is-big-data.html)

A database is considered "big-data" when it reaches a size and complexity that make the use of conventional database management systems impractical for storing, processing, and analyzing the data. These databases typically have billions or trillions of records and require highly specialized hardware and software infrastructure to handle the workload.

The database with 179,702,229 rows used in the project can be considered a large dataset, but not necessarily "big-data," since this amount of records can be managed by a conventional database system on a reasonably powerful computer. However, the size of the database can present challenges for processing and analysis, especially if complex operations are performed on a large scale.

```python
import os
import pandas as pd

pd.options.display.float_format = "{:,.2f}".format
```

```python
df = pd.read_csv("/kaggle/input/ibm-transactions-for-anti-money-laundering-aml/HI-Large_Trans.csv", nrows=5_000)
print(df.columns); display(df)
del df
```

```python
source = "/kaggle/input/ibm-transactions-for-anti-money-laundering-aml/HI-Large_Trans.csv"

df_calculated = pd.DataFrame()
df_len = 0

cols = ["Amount Received", "Receiving Currency", "Amount Paid"]
for i, chunk in enumerate(pd.read_csv(source, chunksize=10_000)): 
    chunk_selection = chunk[cols]
    chunk_group = chunk_selection.groupby(cols[1], as_index=False).sum()
    df_calculated = pd.concat([df_calculated, chunk_group])
    df_len += len(chunk)
    # if i > 5:
    #     break

df_calculated = df_calculated.groupby(cols[1], as_index=False).sum()
```

```python
print(f">> total number of database rows: {df_len:,}")
display(df_calculated)
```
